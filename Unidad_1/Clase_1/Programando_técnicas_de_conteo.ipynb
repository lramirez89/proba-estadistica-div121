# Conteo sin reposición

Vamos a simular un conteo sin reposición en lenguaje Python.

> Un sistema de seguridad requiere configurar una contraseña de longitud $L$. Para armarla, el sistema permite utilizar únicamente los caracteres de un conjunto específico (por ejemplo: 'a', 'b', 'c', 'd', 'e'). Diseñar un algoritmo recursivo que calcule cuántas contraseñas distintas se pueden generar y muestre cómo cambia este número si el sistema permite, o prohíbe, repetir caracteres

1. Con lo que sabemos de las técnicas de conteo, ¿cuántas contaseñas se pueden armar?


```python
def contrasenas_con_repeticion(disponibles, longitud_deseada, actual=""):
    # Caso base: la contraseña alcanzó el tamaño requerido
    if len(actual) == longitud_deseada:
        # print(actual) # (Ojo: para L=5 imprimirá 3125 líneas)
        print(actual)
        return 1

    contador = 0
    # El bucle siempre itera sobre TODOS los caracteres disponibles
    for letra in disponibles:
        # La llamada recursiva recibe el mismo array 'disponibles' intacto
        contador += contrasenas_con_repeticion(disponibles, longitud_deseada, actual + letra)

    return contador

pool_letras = ['a', 'b', 'c', 'd', 'e']
longitud = 5

total_con_repeticion = contrasenas_con_repeticion(pool_letras, longitud)
print(f"Contraseñas posibles (con repetición): {total_con_repeticion}")
# Resultado: 3125 (5^5)
```

#El escenario sin repetición (Backtracking y el Factorial)


2. ¿Qué diferencia hay entre este código y el anterior? ¿Por qué no hay letra repetidas?

```python
def contrasenas_sin_repeticion(disponibles, longitud_deseada, actual=""):
    if len(actual) == longitud_deseada:
        print(actual)
        return 1

    contador = 0
    for letra in disponibles:
        # Filtramos la letra que estamos usando en esta iteración
        restantes = [c for c in disponibles if c != letra]

        # La llamada recursiva ahora recibe un pool reducido
        contador += contrasenas_sin_repeticion(restantes, longitud_deseada, actual + letra)

    return contador

pool_letras = ['a', 'b', 'c', 'd', 'e']
longitud = 5

total_sin_repeticion = contrasenas_sin_repeticion(pool_letras, longitud)
print(f"Contraseñas posibles (sin repetición): {total_sin_repeticion}")
# Resultado: 120 (5!)
```

Los programas anteriores están basados en la técnica de [*backtracking*](https://www.geeksforgeeks.org/dsa/introduction-to-backtracking-2/), común en los problemas computaciones de  [*optimizacion combinatoria*](https://es.wikipedia.org/wiki/Optimizaci%C3%B3n_combinatoria).
